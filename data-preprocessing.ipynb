```python
import json
import re
```

```python
def read_data1():
    with open('testcm.min-zh.min', 'r', encoding='utf-8') as file:
        lines = file.readlines()
        return lines

def read_data2():
    with open('validcm.min-zh.min', 'r', encoding='utf-8') as file:
        lines = file.readlines()
        return lines

def read_data3():
    with open('traincm.min-zh.min', 'r', encoding='utf-8') as file:
        lines = file.readlines()
        return lines

def highlight_marked_unicode(text):
    # Match any character (.) that is followed by _@
    pattern = r'(.)_@'
    # Replace with @@<char>##
    return re.sub(pattern, r'@@\1##', text)
```

```python
with open('preprocessed_data.jsonl', mode='w', newline='', encoding='utf-8') as outfile:
    data = read_data1()
    
    for s in data:
        s = re.sub(r"\s+", "", s) # remove all spaces
        s = highlight_marked_unicode(s) # transform all markers
        s = s.replace("\n", "") # remove new line characters
        r = str(s)
        s = s.replace("#", "") # remove all #
        s = s.replace("@", "") # remove all @
        p = str(s)
        fine_tuning_data = {
            "messages": [
                {"role": "user", "content": p},
                {"role": "assistant", "content": r}
            ]
        }
        
        # Write this dictionary as a JSON object to the output file (JSON Lines format)
        json.dump(fine_tuning_data, outfile, ensure_ascii=False)
        outfile.write('\n')  # Ensure each entry is on a new line
    data2 = read_data2()
    
    for s in data2:
        s = re.sub(r"\s+", "", s) # remove all spaces
        s = highlight_marked_unicode(s) # transform all markers
        s = s.replace("\n", "") # remove new line characters
        r = str(s)
        s = s.replace("#", "") # remove all #
        s = s.replace("@", "") # remove all @
        p = str(s)
        fine_tuning_data = {
            "messages": [
                {"role": "user", "content": p},
                {"role": "assistant", "content": r}
            ]
        }
        
        # Write this dictionary as a JSON object to the output file (JSON Lines format)
        json.dump(fine_tuning_data, outfile, ensure_ascii=False)
        outfile.write('\n')  # Ensure each entry is on a new line
    data3 = read_data3()
    
    for s in data3:
        s = re.sub(r"\s+", "", s) # remove all spaces
        s = highlight_marked_unicode(s) # transform all markers
        s = s.replace("\n", "") # remove new line characters
        r = str(s)
        s = s.replace("#", "") # remove all #
        s = s.replace("@", "") # remove all @
        p = str(s)
        fine_tuning_data = {
            "messages": [
                {"role": "user", "content": p},
                {"role": "assistant", "content": r}
            ]
        }
        
        # Write this dictionary as a JSON object to the output file (JSON Lines format)
        json.dump(fine_tuning_data, outfile, ensure_ascii=False)
        outfile.write('\n')  # Ensure each entry is on a new line
```
