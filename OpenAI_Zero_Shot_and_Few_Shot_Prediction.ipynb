# Use OpenAI Models To Do Zero-Shot and Few-Shot Prediction

- We think `gpt-4o`, `gpt-4o-mini`, and `gpt-3.5-turbo-instruct` are the best models for this project because they are relatively low in inference cost and the last model is instruction tuned.   
- Based on the methods used in [this paper](https://github.com/shuhewang1998/gpt-ner)

## Prompting Methods
- Since the characters we're dealing with is in Traditional Chinese, we think giving the models prompt in Traditional Chinese could enhance the performance.

```python
prompt_cn = "你是一位優秀的中文及臺語語言學家。"
prompt_en = "You are an excellent linguist in Mandarin Chinese and Taiwanese Hokkien."
```

### Zero-Shot Chinese

```python
prompt_zs_cn = prompt_cn + "你的任務是標記給定臺語句子中參雜的中文普通話。"
```

### Few-Shot Chinese

```python
prompt_fs_cn = prompt_zs_cn + """以下是一些範例。

輸入：請上門的客人試食
輸出：請上門的@@客##@@人##試食

輸入：𪜶想欲的不巧一个公道佮尊嚴
輸出：𪜶想欲的@@不##@@巧##一个公道佮尊嚴

輸入：伊的修養真好，攏袂佮人冤家。
輸出：伊的@@修##@@養##真好，攏袂佮人冤家。
"""
```

### Zero-Shot English

```python
prompt_zs_en = prompt_en + " Your task is to label the characters that are in Mandarin Chinese in the given Taiwanese-Hokkien sentence."
```

### Few-Shot English

```python
prompt_fs_en = prompt_zs_en + """ Below are some examples.

input: 請上門的客人試食
output: 請上門的@@客##@@人##試食

input: 𪜶想欲的不巧一个公道佮尊嚴
output: 𪜶想欲的@@不##@@巧##一个公道佮尊嚴

input: 伊的修養真好，攏袂佮人冤家。
output: 伊的@@修##@@養##真好，攏袂佮人冤家。
"""
```

## Self-Verification

### Zero-Shot Chinese

```python
verification_zs_cn = prompt_cn + "你的任務是驗證給定的漢字在臺語參雜中文普通話的句子中是否為中文普通話。"
```

### Few-Shot Chinese

```python
verification_fs_cn = verification_zs_cn + """以下是一些範例。

給定的句子：請上門的客人試食
請問下列漢字在給定的句子中是否爲中文普通話？"客人" 請回答是或否。
是

給定的句子：𪜶想欲的不巧一个公道佮尊嚴
請問下列漢字在給定的句子中是否爲中文普通話？"佮尊嚴" 請回答是或否。
否
"""
```

### Zero-Shot English

```python
verification_zs_en = prompt_en + " Your task is to verify whether the labeled characters are in Mandarin Chinese in the given Taiwanese-Hokkien sentence."
```

### Few-Shot English

```python
verification_fs_en = verification_zs_en + """ Below are some examples.

The given sentence: 請上門的客人試食
Are the characters "客人" in the given sentence Mandarin Chinese? Please answer with yes or not.
Yes

The given sentence: 𪜶想欲的不巧一个公道佮尊嚴
Are the characters "佮尊嚴" in the given sentence Mandarin Chinese? Please answer with yes or not.
No
"""
```

```python
print(prompt_zs_cn)
print(prompt_fs_cn)
print(prompt_zs_en)
print(prompt_fs_en)
print(verification_zs_cn)
print(verification_fs_cn)
print(verification_zs_en)
print(verification_fs_en)
```
